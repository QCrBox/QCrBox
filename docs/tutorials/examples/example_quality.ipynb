# Creation of data quality plots

These are currently created into a folder. This example file is tentative and also used for testing. It will probably not be contained in the final folder, where the involved commands will be used to create parts of the GUI

```python
import shutil

from qcrbox_wrapper import QCrBoxPathHelper, QCrBoxWrapper
```

```python
pathhelper = QCrBoxPathHelper.from_dotenv(".env.dev", "examples_quality/mo_cluster-cifvis")
```

```python
qcrbox = QCrBoxWrapper.from_server_addr("127.0.0.1", 11000)
```

```python
quality = qcrbox.application_dict['QCrBox Quality']
```

```python
original_cif = pathhelper.path_to_local("input.cif")
```

```python
local_basic_model_path, qcrbox_basic_model_path = pathhelper.path_to_pair("basic_model_quality_indicators")
local_basic_model_path.mkdir(exist_ok=True)
shutil.copy(original_cif, local_basic_model_path / "input.cif")
calc = quality.basic_model_quality_indicators(
    input_cif_path=qcrbox_basic_model_path / "input.cif",
    output_html_path= qcrbox_basic_model_path / "basic_indicators.html"
)
calc.wait_while_running(0.2)
```

```python
local_fo_fc_plotly_path, qcrbox_fo_fc_plotly_path = pathhelper.path_to_pair("fo_fc_plotly")
local_fo_fc_plotly_path.mkdir(exist_ok=True)
shutil.copy(original_cif, local_fo_fc_plotly_path / "input.cif")
calc = quality.fobs_div_fcalc_plotly(
    input_cif_path=qcrbox_fo_fc_plotly_path / "input.cif",
    output_html_path= qcrbox_fo_fc_plotly_path / "fo_fc_plotly.html"
)
calc.wait_while_running(0.2)
```

```python
local_fo_fc_bokeh_path, qcrbox_fo_fc_bokeh_path = pathhelper.path_to_pair("fo_fc_bokeh")
local_fo_fc_bokeh_path.mkdir(exist_ok=True)
shutil.copy(original_cif, local_fo_fc_bokeh_path / "input.cif")
calc = quality.fobs_div_fcalc_bokeh(
    input_cif_path=qcrbox_fo_fc_bokeh_path / "input.cif",
    output_html_path= qcrbox_fo_fc_bokeh_path / "fo_fc_bokeh.html"
)
calc.wait_while_running(0.2)
```

```python
local_ortep_path, qcrbox_ortep_path = pathhelper.path_to_pair("ortep")
local_ortep_path.mkdir(exist_ok=True)
shutil.copy(original_cif, local_ortep_path / "input.cif")
calc = quality.ortep_3d(
    input_cif_path=qcrbox_ortep_path / "input.cif",
    output_html_path= qcrbox_ortep_path / "ortep.html"
)
calc.wait_while_running(0.2)
```

```python
local_ortep_path, qcrbox_ortep_path = pathhelper.path_to_pair("ortep_cifvis")
local_ortep_path.mkdir(exist_ok=True)
shutil.copy(original_cif, local_ortep_path / "input.cif")
calc = quality.ortep_cifvis_3d(
    input_cif_path=qcrbox_ortep_path / "input.cif",
    output_html_path= qcrbox_ortep_path / "ortep.html"
)
calc.wait_while_running(0.2)
```
